```python
from shared.model_utils import load_model, get_next_token_probs, compute_entropy

model, tokenizer = load_model(quantize=True)  # 4-bit for Kaggle's 16GB GPUs
print("Model loaded.")
```

```python
confident_prompt = "The capital of France is"
ambiguous_prompt = "Tomorrow we should go to the"

p1 = get_next_token_probs(model, tokenizer, confident_prompt)
p2 = get_next_token_probs(model, tokenizer, ambiguous_prompt)

e1 = compute_entropy(p1)
e2 = compute_entropy(p2)

print(f"Confident prompt entropy: {e1:.3f}")
print(f"Ambiguous prompt entropy: {e2:.3f}")

assert e2 > e1, "Sanity check failed — ambiguous prompt should have higher entropy. Investigate before proceeding."
print("\nSanity check passed. Environment is ready.")
```
